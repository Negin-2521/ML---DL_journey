# Modeling with Logistci Regression 
## Using Artificial Datasets

### Importing libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_blobs
from sklearn.linear_model import LogisticRegression, SGDClassifier
```

### Creating Dataset

```python
A_dataset=make_blobs(n_samples=400, centers=3 ,n_features=2, cluster_std=2.5, random_state=42)
```

```python
A_dataset_f=np.array(A_dataset[0])
A_dataset_l=np.array(A_dataset[1])
```

```python
x_train,x_test,y_train, y_test=train_test_split(A_dataset_f, A_dataset_l, test_size=0.2, random_state=45)
```

```python
plt.scatter(x_train[:,0], x_train[:,1], c=y_train)
plt.show()
```

#### Logistci Regression

```python
LR_model=LogisticRegression(max_iter=1000, random_state=42)
LR_model.fit(x_train, y_train)
```

```python
train_accuracy_lr=LR_model.score(x_train, y_train)
test_accuracy_lr=LR_model.score(x_test, y_test)

print(f'Train Dataset Accuracy is : {train_accuracy_lr:.3f}')
print(f'Test Dataset Accuracy is : {test_accuracy_lr:.3f}')
```

#### SGD Classifier

```python
sgd_model=SGDClassifier(loss='log_loss', random_state=42)
sgd_model.fit(x_train, y_train)
```

```python
train_accuracy_sgd=sgd_model.score(x_train, y_train)
test_accuracy_sgd=sgd_model.score(x_test, y_test)

print(f'Train Dataset Accuracy is : {train_accuracy_sgd:.3f}')
print(f'Test Dataset Accuracy is : {test_accuracy_sgd:.3f}')
```

```python
result=pd.DataFrame({'Algorithm':['Logistic regression','SGD Classifier'],
                    'Train Accuracy':[round(train_accuracy_lr,3),round(train_accuracy_sgd,3)],
                    'Test Accuracy':[round(test_accuracy_lr,3),round(test_accuracy_sgd,3)]})
print(result)
```
